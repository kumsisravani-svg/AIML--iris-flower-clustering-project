```python
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.cluster import KMeans
```

```python
df = pd.read_csv("archive (10)/Iris.csv")

df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
X = df.iloc[:, 1:5]

X.head()
```

```python
kmeans = KMeans(n_clusters=3, random_state=42)

kmeans.fit(X)
```

```python
df["Cluster"] = kmeans.labels_

df.head()
```

```python
plt.figure(figsize=(8,5))

plt.scatter(
    df["SepalLengthCm"],
    df["PetalLengthCm"],
    c=df["Cluster"]
)

plt.xlabel("Sepal Length")
plt.ylabel("Petal Length")

plt.title("Iris Flower Clusters")

plt.show()
```

```python
comparison = df[["Species", "Cluster"]]

comparison.head(10)
```

```python
df.to_csv("cluster_output.csv", index=False)

print("Cluster output saved successfully!")
```
